```python
#C1 & C2 : Fix the Dtypes and handle the missing values 
#Convert ride_date to datetime ad then verify 

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df = pd.read_csv('../data/citibike_weather_daily.csv') #Loading our data set into the preprocessing folder 

df['ride_date'] = pd.to_datetime(df['ride_date'])

df.dtypes #verifying that ride_date has been converted properly
```

```python
df['precip_in'] = df['precip_in'].replace(99.99, np.nan) #replaced the sentinel value 99.99 with NaN
df['precip_in'].describe()  #verifying the distribution

```

```python
df['precip_in'].isna().sum() #only 1 missing percipitation day in the entire dataset
```

```python
df['precip_in'] = df['precip_in'].fillna(0) # fillin in the pne missing day as zero 

##replaced the NOAA GSOD sentinel value 99.99 with NaN and found only one missing precipitation value.
##Because precipitation is typically zero on most days and GSOD outages often occur on dry days, I chose to impute missing precipitation as 0.
##This preserves the full dataset and avoids dropping rows unnecessarily. If there were more rows, I would be likely to drop the missing rows if it still would lead to preserving the data accuracy
```

```python
#C3: Encode the Days of the week 
day_map = {
    'Monday': 0,
    'Tuesday': 1,
    'Wednesday': 2,
    'Thursday': 3,
    'Friday': 4,
    'Saturday': 5,
    'Sunday': 6
}

df['day_of_week_encoded'] = df['day_of_week'].map(day_map)

df[['day_of_week', 'day_of_week_encoded']].head()  #verifying values 
```

```python
#C4:Give your model a way to know about the system growth you found in E4. Create either a year column or a days-since-launch index (or both, and pick one for modeling).
df['days_since_launch'] = (df['ride_date'] - df['ride_date'].min()).dt.days

#Subtracted the  earliest ride_date from each row to measure how many days have passed since the system launched. Using .dt.days converts that time difference into a numeric count > ".dt" datetime within the ".days" is a datetime attribute being used
```

```python
#Saving my finished dataframe to data/citibike_weather_daily_clean.csv. This file is what model.ipynb loads — the raw CSV's job is done.


df.to_csv('../data/citibike_weather_daily_clean.csv', index=False)

#index = False b/c we do not want panda to an extra column on row #'s to the CSV. 
```
